### Preparación: librerías, partición y funciones de métricas

````python
import numpy as np
import matplotlib.pyplot as plt
import warnings
from statsmodels.tsa.seasonal import seasonal_decompose
from statsmodels.tsa.holtwinters import ExponentialSmoothing

warnings.filterwarnings("ignore")
plt.rcParams["figure.figsize"] = (11, 4)

# Partición fija
train = hidro[:"2025-06"]
test = hidro["2025-07":"2026-06"]

# Métricas
def rmse(y_true, y_pred):
    y_true, y_pred = np.asarray(y_true, float), np.asarray(y_pred, float)
    return np.sqrt(np.mean((y_true - y_pred) ** 2))

def mae(y_true, y_pred):
    y_true, y_pred = np.asarray(y_true, float), np.asarray(y_pred, float)
    return np.mean(np.abs(y_true - y_pred))

def mase(y_true, y_pred, y_train, m=12):
    """MAE del pronóstico / MAE del ingenuo estacional (rezago m) dentro del entrenamiento."""
    y_train = np.asarray(y_train, float)
    escala = np.mean(np.abs(y_train[m:] - y_train[:-m]))
    return mae(y_true, y_pred) / escala

def metricas(y_true, y_pred, y_train):
    return {"RMSE": rmse(y_true, y_pred),
            "MAE": mae(y_true, y_pred),
            "MASE": mase(y_true, y_pred, y_train)}

print(train.shape, test.shape)
print(f"Entrenamiento: {train.index[0]:%Y-%m} a {train.index[-1]:%Y-%m} | "
      f"Prueba: {test.index[0]:%Y-%m} a {test.index[-1]:%Y-%m}")
````

## 1.b. Un modelo clásico (12 pts)

**(I)** Ajustar Holt-Winters o SARIMA (statsmodels) sobre el entrenamiento. Justificar la configuración: ¿tendencia sí o no?, ¿qué órdenes?, ¿constante?

````python
def ajustar_hw(y_tr):
    """Holt-Winters: estacionalidad aditiva (m = 12), sin tendencia."""
    return ExponentialSmoothing(y_tr, trend=None, seasonal="add", seasonal_periods=12,
                                initialization_method="estimated").fit()

fit_hw = ajustar_hw(train)
print("Holt-Winters aditivo sin tendencia, ajustado sobre", train.index[0].strftime("%Y-%m"),
      "a", train.index[-1].strftime("%Y-%m"))
print(f"alpha (nivel)        = {fit_hw.params['smoothing_level']:.3f}")
print(f"gamma (estacional)   = {fit_hw.params['smoothing_seasonal']:.3f}")
print(f"nivel inicial        = {fit_hw.params['initial_level']:.1f} GWh")
print(f"AICc = {fit_hw.aicc:.1f}   SSE = {fit_hw.sse:.0f}")
````

**Justificación de la configuración:**

- **Modelo:** Holt-Winters con estacionalidad aditiva y periodo 12.
- **¿Tendencia? No.** La tendencia de la descomposición cambia de dirección (sube 2019–2022, baja en 2023, sube 2024–2025). Esos cambios vienen de la hidrología (El Niño), no de un crecimiento estructural y la generación hidro está acotada por la capacidad instalada.
- **¿Aditiva o multiplicativa?** Aditiva. La amplitud seca–lluviosa ronda 300–400 GWh tanto en años de nivel bajo (2019, 2024) como alto (2021, 2025); no crece proporcionalmente con el nivel.
- **¿Constante?** Holt-Winters no tiene un término constante aparte: el nivel Lt actúa como un intercepto que cambia en el tiempo. En un SARIMA con diferencia estacional, una constante implicaría una deriva, que se descarta por la misma razón que la tendencia.

**(II)** Pronosticar los 12 meses de prueba, calcular RMSE, MAE y MASE, y graficar observado vs. pronosticado.

````python
pred_hw = fit_hw.forecast(12)
met_hw = metricas(test, pred_hw, train)

print(pd.DataFrame({"observado": test, "Holt-Winters": pred_hw.round(1)}))
print("\nHolt-Winters (prueba jul-2025 a jun-2026):")
for k, v in met_hw.items():
    print(f"  {k}: {v:.3f}")
````

````python
import logging
from prophet import Prophet
logging.getLogger("cmdstanpy").setLevel(logging.WARNING)
logging.getLogger("prophet").setLevel(logging.WARNING)

def ajustar_prophet(y_tr, h=12, interval_width=0.8, uncertainty_samples=1000):
    """Prophet con estacionalidad anual aditiva. Con datos mensuales solo hay 12 fases por año,
    así que se usan 6 términos de Fourier (el máximo que distinguen 12 puntos)."""
    df = pd.DataFrame({"ds": y_tr.index, "y": y_tr.values})
    m = Prophet(yearly_seasonality=6, weekly_seasonality=False, daily_seasonality=False,
                seasonality_mode="additive", interval_width=interval_width,
                uncertainty_samples=uncertainty_samples)
    m.fit(df)
    futuro = m.make_future_dataframe(periods=h, freq="MS")
    fc = m.predict(futuro).tail(h).set_index("ds")
    return m, fc

m_pr, fc_pr = ajustar_prophet(train)
pred_pr = pd.Series(fc_pr["yhat"].values, index=test.index, name="Prophet")
print(fc_pr[["yhat", "yhat_lower", "yhat_upper"]].round(1))
````

## 2.1. Regularización (6 pts)

**(a) ¿Qué hacen Ridge y Lasso sobre los coeficientes y por qué ayudan a controlar el sobreajuste?**

**Respuesta:** Ridge y Lasso agregan una penalización al ajuste por mínimos cuadrados para reducir la magnitud de los coeficientes.
* Ridge penaliza la suma de sus cuadrados (L2 — Ridge (cuadrados)),
* Lasso la suma de sus valores absolutos (L1 — Lasso (valor absoluto)).

Así evitan que el modelo se ajuste demasiado a las particularidades o al ruido del entrenamiento, lo que puede mejorar su generalización a meses no observados.



**(b) Una diferencia práctica entre Ridge y Lasso, relacionada con el compromiso sesgo–varianza.**

**Respuesta:** Ridge y Lasso ayudan a controlar el sobreajuste, pero lo hacen de maneras diferentes. Ridge reduce progresivamente la magnitud de los coeficientes sin eliminarlos normalmente, por lo que resulta útil cuando varias variables aportan información al modelo, incluso si están correlacionadas. Por su parte, Lasso puede reducir algunos coeficientes exactamente a cero, permitiendo seleccionar las variables más relevantes y simplificar el modelo.

* Ridge suele reducir los coeficientes sin llevarlos exactamente a cero;
* Lasso puede hacer algunos iguales a cero y, por tanto, seleccionar variables.
* Ambos introducen algo de sesgo a cambio de reducir la varianza del modelo. Si la penalización es excesiva, aumenta demasiado el sesgo y se pierde capacidad predictiva.

**(c) ¿Cómo es posible que Ridge y Lasso dieran mayor RMSE de prueba que mínimos cuadrados? ¿Por qué se validó con *TimeSeriesSplit* y no con K-fold común?**

**Respuesta:** La regularización no garantiza mejorar el RMSE de prueba: si mínimos cuadrados ya generaliza bien, o si la penalización elegida es demasiado fuerte, Ridge y Lasso pueden introducir sesgo adicional y empeorar las predicciones
* Se utiliza *TimeSeriesSplit* porque preserva el orden cronológico: entrena con datos pasados y valida con datos posteriores.
* Un K-fold común puede entrenar con observaciones futuras respecto de los meses evaluados y dar una estimación demasiado optimista.

En consecuencia, *TimeSeriesSplit* permite evaluar los modelos de forma más realista y seleccionar el nivel de regularización considerando su desempeño sobre períodos futuros, sin utilizar información que todavía no estaría disponible al momento de realizar la predicción.

## 2.2. Cadena de Markov con datos reales (7 pts)

$$P = \begin{pmatrix} 0.83 & 0.17 \\ 0.29 & 0.71 \end{pmatrix}$$

(A: hidro ≥ 700 GWh, B: hidro baja; filas = mes actual, columnas = mes siguiente)

**(a) Interpretar en palabras la segunda fila.**

**Respuesta:** 
* La segunda fila corresponde al estado B (generación hidroeléctrica inferior a 700 GWh).
* Si el mes actual está en B, hay 29 % de probabilidad de pasar a A el mes siguiente y 71 % de probabilidad de permanecer en B.

**(b) ¿Qué representa la distribución estacionaria? Plantear $\pi P = \pi$ con $\pi_A + \pi_B = 1$ y resolver.**

**Respuesta:** La distribución estacionaria representa la proporción de tiempo que una cadena de Markov pasa en cada estado a largo plazo, suponiendo que las probabilidades de transición permanecen constantes. En este caso, permite estimar qué porcentaje de meses se encontraría en los estados A y B una vez que el sistema alcanza su equilibrio.

Para calcularla, se utiliza la condición:

$\pi=(\pi_A,\pi_B)$ se plantea $\pi P=\pi$ y $\pi_A+\pi_B=1$.

De la primera componente:

$\pi_A=0.83\pi_A+0.29\pi_B$, por lo que $0.17\pi_A=0.29\pi_B$.

Así:

$\pi_A=\frac{0.29}{0.46}\approx0.6304$

por otro lado:

$\pi_B=\frac{0.17}{0.46}\approx0.3696$.

A largo plazo serían aproximadamente **63.0 % de meses A y 37.0 % de meses B**, según esta cadena (*0.63 + 0.37 = 1*).

Es importante destacar que estos porcentajes no indican que los estados se alternarán siguiendo una frecuencia exacta, sino que representan las proporciones esperadas a largo plazo. Además, al alcanzar la distribución estacionaria, las probabilidades globales de estar en cada estado se mantienen constantes de un período al siguiente, aunque individualmente los meses puedan cambiar de estado.

**(c) ¿Es razonable el supuesto de Markov para la hidrología de un país con años de El Niño? Argumentar.**

**Respuesta:** El supuesto de Markov es razonable como una primera aproximación, pero tiene limitaciones para representar el comportamiento de la generación hidroeléctrica de Costa Rica. Este modelo considera que el estado del siguiente mes depende únicamente del estado actual, cuando en realidad también pueden influir *factores como la estacionalidad, las lluvias acumuladas, los niveles de los embalses y fenómenos climáticos persistentes como El Niño.*

Por lo tanto, aunque la cadena de Markov permite identificar patrones de transición entre meses, sería conveniente incorporar la estacionalidad u otras variables relevantes para representar mejor el comportamiento del sistema, especialmente durante períodos de variabilidad climática.

**(d) En una frase, conectar la cadena de Markov con PageRank o con los modelos de lenguaje.**

**Respuesta:** PageRank utiliza cadenas de Markov de forma similar a nuestro modelo de generación hidroeléctrica: mientras nuestro modelo calcula las probabilidades de transición entre estados de generación alta y baja, PageRank calcula las probabilidades de navegar entre páginas web y utiliza la distribución estacionaria para determinar su importancia relativa.

## 2.3. Leer código (6 pts)

```python
import numpy as np
rng = np.random.default_rng(2026)
def log_post(p):  # 18 exitos en 50 intentos, prior plano
    return 18*np.log(p) + 32*np.log(1-p) if 0 < p < 1 else -np.inf
p, cadena = 0.5, []
for _ in range(20000):
    prop = p + rng.normal(0, 0.1)
    if np.log(rng.uniform()) < log_post(prop) - log_post(p):
        p = prop
    cadena.append(p)
print(np.mean(cadena[2000:]))
```

**(a) ¿Monte Carlo o MCMC? ¿Qué distribución muestrea y qué valor aproximado imprime? Justificar.**

**Respuesta:** Es **MCMC**, específicamente Metropolis-Hastings, porque construye una cadena de propuestas que acepta o rechaza para muestrear una distribución posterior.
* Con 18 éxitos de 50 y una previa uniforme $\mathrm{Beta}(1,1)$,
* la posterior es $\mathrm{Beta}(19,33)$; su media es $19/52\approx0.3654$.
* Por eso el promedio impreso debería estar cerca de **0.365**, aunque variará ligeramente por el muestreo.

**(b) Papel de *prop*, de la comparación con *rng.uniform()* y del descarte de las primeras 2000 iteraciones.**

**Respuesta:** *prop* es el nuevo valor candidato de $p$, obtenido al sumar una perturbación normal al estado actual. La comparación de logaritmos con *rng.uniform()* implementa la regla de aceptación de Metropolis-Hastings: acepta propuestas favorecidas por la posterior y algunas menos favorecidas con cierta probabilidad.
* Se descartan las primeras 2 000 iteraciones como **calentamiento** (*burn-in*) para reducir la influencia del valor inicial $p=0.5$.
* De esta manera, el promedio se calcula utilizando las 18 000 iteraciones restantes, buscando obtener una estimación más representativa de la distribución posterior.

**(c) En dos frases, diferencia entre la simulación Monte Carlo de 1.e(III) y este código.**

**Respuesta:**
* En Monte Carlo simula muchos escenarios futuros de generación hidroeléctrica para estimar la probabilidad de estar por debajo de 550 GWh en marzo de 2027.
* En cambio, este código MCMC genera muestras **dependientes** de la distribución posterior de una probabilidad desconocida, usando propuestas y una regla de aceptación.

## 2.4. Leer resultados (6 pts)

**(a) MASE promedio 1.10 (mejor modelo) vs. 1.40 (ingenuo estacional): ¿es malo el modelo por tener MASE > 1? ¿Qué compara exactamente el MASE?**

**Respuesta:** No necesariamente: el modelo tiene MASE promedio de **1.10**, menor que el **1.40** del ingenuo estacional en esas mismas ventanas, así que fue mejor en esa comparación.
* El MASE divide el MAE del pronóstico por el MAE de un pronóstico ingenuo estacional calculado **dentro del entrenamiento** (con rezago de 12 meses).
* Un MASE mayor que 1 indica que el error supera esa escala histórica, pero no implica que sea peor que el ingenuo estacional evaluado sobre el mismo período de prueba.

En este caso, ambos modelos superaron la escala de error histórica, pero el modelo seleccionado presentó un menor error relativo, lo que demuestra un mejor desempeño predictivo en las ventanas evaluadas.

**(b) ¿Qué significa que Chronos-2 sea zero-shot? Una situación donde es ventaja y otra donde no basta.**
**Respuesta:**

*Zero-shot* significa que Chronos-2 puede generar pronósticos de una serie nueva sin entrenarse específicamente con esa serie.
* Es útil cuando se necesita un pronóstico inicial rápido o hay pocos recursos para ajustar un modelo local; sin embargo, puede no bastar cuando la serie presenta cambios excepcionales, como una sequía intensa, o condiciones locales que el modelo no representa bien y que requieren validación y ajustes.

**(c) Con un año de prueba ganaba SARIMA; con tres ventanas, Holt-Winters. ¿Qué enseña sobre cómo elegir un modelo?**

**Respuesta:** Un único año de prueba puede favorecer a un modelo por condiciones particulares de ese período. Evaluar varias ventanas expansivas permite comprobar la estabilidad del desempeño bajo distintos contextos, incluido 2024.

Por lo tanto, la selección del modelo debe basarse en su capacidad de generalización y en su desempeño a través de diferentes períodos, en lugar de elegirlo únicamente por obtener el menor error en un año específico.
